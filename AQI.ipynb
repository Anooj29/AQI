```python
features = df2[['City', 'PM2.5', 'PM10', 'NO', 'NO2', 'NOx', 'NH3', 'CO', 'SO2', 'O3',
       'Benzene', 'Toluene', 'Xylene']]

labels = df2['AQI']
```

```python
features
```

```python
labels
```

```python
from sklearn.model_selection import train_test_split
```

```python
Xtrain, Xtest, Ytrain, Ytest = train_test_split(features, labels, test_size=0.2, random_state=2)
```

```python
from sklearn.metrics import classification_report
from sklearn import metrics
from sklearn import tree 
```

```python
Xtrain = pd.get_dummies(Xtrain)
Xtest = pd.get_dummies(Xtest)

# Align the test set columns with train
Xtest = Xtest.reindex(columns=Xtrain.columns, fill_value=0)
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.datasets import make_regression 
regr = RandomForestRegressor(max_depth=2, random_state=0)
regr.fit(Xtrain, Ytrain)
print(regr.predict(Xtest))
```

```python
y_pred = regr.predict(Xtest)
```

```python
from sklearn.metrics import r2_score
r2_score(Ytest, y_pred)
```

```python

def predict_aqi():
  
    city = input("Enter City: ")
    pm25 = float(input("Enter PM2.5 value: "))
    pm10 = float(input("Enter PM10 value: "))
    no = float(input("Enter NO value: "))
    no2 = float(input("Enter NO2 value: "))
    nox = float(input("Enter NOx value: "))
    nh3 = float(input("Enter NH3 value: "))
    co = float(input("Enter CO value: "))
    so2 = float(input("Enter SO2 value: "))
    o3 = float(input("Enter O3 value: "))
    benzene = float(input("Enter Benzene value: "))
    toluene = float(input("Enter Toluene value: "))
    xylene = float(input("Enter Xylene value: "))

   
    input_data = pd.DataFrame({
        'City': [city],
        'PM2.5': [pm25],
        'PM10': [pm10],
        'NO': [no],
        'NO2': [no2],
        'NOx': [nox],
        'NH3': [nh3],
        'CO': [co],
        'SO2': [so2],
        'O3': [o3],
        'Benzene': [benzene],
        'Toluene': [toluene],
        'Xylene': [xylene]
    })

   
    input_data = pd.get_dummies(input_data)
    input_data = input_data.reindex(columns=Xtrain.columns, fill_value=0)

    
    prediction = regr.predict(input_data)
    print(f"Predicted AQI: {prediction[0]}")

predict_aqi()
```
